## Looking at data
### Matplotlib: https://matplotlib.org/
Although we will focus on plotting later, for now let's introduce the matplotlib package in order to see our images and start visualizing something!

```python
import numpy as np
from matplotlib import pyplot as plt # pyplot is the core plotting package of matplotlib, it is a convention to nickname it plt

# most things can be done with the plt.something syntax
# later we will use a lot of plt.plot and plt.scatter

# let's check that the uniform distribution of Numpy is actually normally distributed
normal_array = np.random.normal(loc=0, scale=2, size=1000) # this should be a normal distribution with mean = 0 and std = 2
plt.hist(normal_array)
plt.show() # it's generally good to have plt.show to see the plot without problems
```

Let's create and visualize an RGB image entirely using Numpy!

```python
rgb_img = np.zeros((10,10,3), dtype=np.uint8) # we create a 10px by 10px little 8-bit image
print(rgb_img.shape) # you can check the shape is correct!

# using slicing we can assing specific values to any elements of the array
     # row, col, layer
rgb_img[2:6, 2:6, 0] = 255# 255 is the max possible value for a 8-bit image!
rgb_img[4:7, 4:7, 1] = 255
rgb_img[5:8, 1:5, 2] = 255
plt.imshow(rgb_img)
plt.show()
```

```python
# just for fun let's create a cooler image!
x = np.linspace(-5, 5, 101)
y = np.linspace(-5, 5, 101)

xx, yy = np.meshgrid(x, y) # what does meshgrid do? A good occasion to use the help function! # help(np.meshgrid)
zz = np.sqrt(xx**2 + yy**2)

plt.imshow(zz)
plt.colorbar() # this adds a colobar for reference to the current open plot
plt.show()
```

```python
fig, axes = plt.subplots(2, 3, sharey=True, sharex=True, figsize=(10,6))
ax = axes.ravel()
i=0
for panel in ax:
    panel.set_title(f"panel_{i}")
    if i<1:
        panel.imshow(zz)
    i=i+1
plt.show()
```

This means that when a virtual environment is deleted, the rest of the system will remain as is: untouched. We will mostly use anaconda / miniconda to create conda virtual environment. Those environment will provide the kernel that Jupyter can use to run your notebooks.
<span style="color: blue">A virtual environment is a directory containing a stand-alone installation of Python (independent from that of the operating system). You can use all basics mathematical operators with numberical variables. Keep in mind that if you are combining integers with floats you will get a float.

## Lists []
It is often very useful to have a collection of any objects instead of using or working on objects one by one. Lists in Python are **ordered** and **mutable**. Lists can also be concatenated together with the sum operator.

````python
list1 = [1,2,3,4,5]
list2 = ['this', 'is', 'a', 'list']
list3 = [None, True, False, 'text', 512.99]
list4 = []

print(list1 + list3)
````

You can modify or access elements of lists using the corresponding index. You can also slice parts of lists using the `:` operator. You can also refer to the last element of a list using `-1`.

To add an element to a list, you can use the `append` function.

````python
print(list1[1])
print(list2[2:4])
print(list3[-1])
list3[0] = 'Yes'
print(list3)
list3.append('New Element')
print(len(list3)) # print the number of elements in a list
````

## Dictionaries {}
Dictionaries are another type of collection of objects characterized by **keys** &rarr; **values** pairs. For every key there is a value. Keys and values can be **anything** but keys have to be unique and dictionaries are not *ordered*.

````python
dictionary1 = {
    'a': 5,
    'b': ['list', 'inside', 'a', 'dictionary'],
    2: True,
    100: None
}
````

You can define new key - value pair or modify an existing one with the syntax below.

You can access keys and values or both using the corresponding functions: `keys()`, `values()` or `items()`

````python
dictionary1['c'] = 'another value'
dictionary1[2] = False

print(dictionary1)
print(dictionary1.keys())
print(dictionary1.values())
print(dictionary1.items())
````

## IF, ELSE, ELIF
Conditionals are used when you want to execute something **only when a specific condition** occurs.

````python
big_number = 5

if big_number > 10:
    print('This number if larger than 10!')
elif big_number < 0:
    print('This number is negative!')
else:
    print('This number is smaller than 10 but it is not negative!')
````

## LOOPS
Very often you don't want to operate on single objects manually. It's much easier and better to use loops and do something multiple times!

You can loop through any **iterable** in Python: usually they are lists, range or other arrays (see later).

### FOR loops
For loops are executed for **every given element** of something.

````python
my_list = ['first element', 'second element', 'third element']
for element in my_list:
    print('This is the', element)
````

````python
for index in range(len(my_list)):
    print(my_list[index])
````

## WHILE loops
While loops keep running **as long as** a condition is met.
<span style="color: blue">We can determine how many elements are in 'my_string' with **len**</span>

````python
my_string = "Oh no! I'm disappearing!"

while len(my_string) > 0:
    print(my_string)
    my_string = my_string[:-4]
````

## Functions
When you have a _package_ or _block_ of code that you have to re-use multiple times, it's time to use functions!

````python
def greatest_number(num1, num2):
    if num1 > num2:
        return num1
    elif num1 < num2:
        return num2
    elif num1 == num2:
        return None
    else: print("Something is wrong!")

list_of_numbers1 = [1e5, -1, 1e2, 1e-5]
list_of_numbers2 = [1e5, -1e5, 10000, 0.1]

# Use zip to loop across elements of multiple iterators synchronously!
for num1, num2 in zip(list_of_numbers1, list_of_numbers2):
    print(greatest_number(num1, num2))
````

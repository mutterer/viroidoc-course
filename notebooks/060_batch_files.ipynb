# Exercise 1

1. loop through all images in data/protein_atlas
2. check if TIF
3. if TIF: read the image
4. save first channel as PNG

```python
from skimage.io import imread, imsave
import os
#from tifffile import imwrite

list_of_files = os.listdir('../data/protein_atlas')

for filename in list_of_files:
    if filename.endswith('.tif'):
        print('Processing file:', filename)
        img = imread('../data/protein_atlas/' + filename)
        png_filename = filename.replace('tif','png')
        imsave('../data/protein_atlas/' + png_filename, img[:,:,0])
        #imwrite('../data/protein_atlas/' + png_filename, img)
        
```

# Exercise 2
1. Loop through all images in data/protein_atlas
2. Extract WELL from filename
3. create a list of unique WELLs

```python
list_of_files = os.listdir('../data/protein_atlas')

well_list = []

for file in list_of_files:
    if file.endswith('.tif'):
        well = file.split('_')[1]
        well_list.append(well)
        
# convert the list to set
# then to list again
# then sort it
unique_wells = sorted(list(set(well_list)))
print(unique_wells)
```
